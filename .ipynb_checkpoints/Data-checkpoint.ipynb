```python
import os
folder = "MyFolder"
filename = "hello.txt"
if os.path.exists(folder):
    print("Folder already exists")
else:
    os.mkdir(folder)
    print("Folder created")
file = folder + "/" + filename
f = open(file, "w")
f.write("Hello, this is my first file.")
f.close()
print("File created successfully")
f = open(file, "r")
print("\nFile Content:")
print(f.read())
f.close()
```
