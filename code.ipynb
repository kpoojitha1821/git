```python
from datetime import datetime, timedelta
total_seats = 10
bookedseats = 0
seats = {}
while True:
    try:
        view_n = int(input("Enter n for View Booking (hours): "))
        booking_n = int(input("Enter n for New Booking (hours): "))
        cancel_n = int(input("Enter n for Cancellation (hours): "))
        if view_n < 0 or booking_n < 0 or cancel_n < 0:
            print("n values must be 0 or greater.")
            continue
        break
    except ValueError:
        print("Please enter numbers only.")
while True:
    try:
        travel_input = input(
            "Enter travel date and time (DD-MM-YYYY HH:MM): "
        )

        travel_date = datetime.strptime(
            travel_input, "%d-%m-%Y %H:%M"
        )

        if travel_date <= datetime.now():
            print("Travel date and time must be in the future.")
            continue
        break
    except ValueError:
        print("Invalid date/time format.")
        print("Example: 02-10-2026 12:20")
print("\nTravel Date/Time:",
      travel_date.strftime("%d-%m-%Y %I:%M %p"))
print("View Booking n:", view_n, "hours")
print("New Booking n:", booking_n, "hours")
print("Cancellation n:", cancel_n, "hours")
while True:
    print("\n1. New Booking")
    print("2. Cancel Booking")
    print("3. View Booking")
    print("0. Quit")
    try:
        choice = int(input("Enter your choice: "))
    except ValueError:
        print("Invalid choice. Please enter numbers only.")
        continue
    if choice == 1:
        current_time = datetime.now()
        booking_limit = current_time + timedelta(hours=booking_n)
        if booking_limit > travel_date:
            print("\nNew booking is not allowed.")
            print("New Booking n:", booking_n, "hours")
            print("Travel date/time:",
                  travel_date.strftime("%d-%m-%Y %I:%M %p"))
            continue
        if bookedseats == total_seats:
            print("All seats are already booked.")
            continue
        seatno = bookedseats + 1
        print("\nNext available seat:", seatno)
        name = input("Enter your name: ").strip()
        if not name:
            print("Invalid name. Please enter a valid name.")
            continue
        if not all(ch.isalpha() or ch == " " for ch in name):
            print("Invalid name. Please enter letters only.")
            continue
        clean_name = name.replace(" ", "").lower()
        if len(clean_name) < 3:
            print("Invalid name. Please enter a valid name.")
            continue
        vowels = "aeiou"
        if not any(ch in vowels for ch in clean_name):
            print("Invalid name. Please enter a valid name.")
            continue
        consonants = 0
        invalid_name = False
        for ch in clean_name:
            if ch not in vowels:
                consonants += 1
            else:
                consonants = 0
            if consonants >= 3:
                invalid_name = True
                break
        if invalid_name:
            print("Invalid name. Please enter a valid name.")
            continue
        try:
            age = int(input("Enter your age: "))
        except ValueError:
            print("Invalid age. Please enter numbers only.")
            continue
        if age < 18 or age > 120:
            print("Invalid age.")
            continue
        phone = input("Enter your phone number: ").strip()
        if not phone.isdigit():
            print("Invalid phone number. Enter numbers only.")
            continue
        if len(phone) != 10:
            print("Invalid phone number. Enter exactly 10 digits.")
            continue
        if phone[0] not in "6789":
            print("Invalid phone number,enter starting number 6,7,8,9 only.")
            continue
        if len(set(phone)) == 1:
            print("Invalid phone number.")
            continue
        seats[seatno] = {
            "name": name,
            "age": age,
            "phone": phone
        }
        bookedseats += 1
        print("\nBooking successful!")
        print("Seat Number:", seatno)
        print("Travel Date/Time:",
              travel_date.strftime("%d-%m-%Y %I:%M %p"))
        print("Available Seats:",
              total_seats - bookedseats)
    elif choice == 2:
        if bookedseats == 0:
            print("No booking available to cancel.")
            continue
        current_time = datetime.now()
        cancel_limit = current_time + timedelta(hours=cancel_n)
        if cancel_limit > travel_date:
            print("\nCancellation is not allowed.")
            print("Cancellation n:", cancel_n, "hours")
            print("Travel date/time:",
                  travel_date.strftime("%d-%m-%Y %I:%M %p"))
            continue
        try:
            seatno = int(input("Enter your seat number: "))
        except ValueError:
            print("Invalid seat number.")
            continue
        if seatno < 1 or seatno > total_seats:
            print("Invalid seat number.")
            continue
        if seatno not in seats:
            print("This seat is not booked.")
            continue
        phone = input(
            "Enter your registered phone number: "
        ).strip()
        if phone != seats[seatno]["phone"]:
            print("Invalid phone number.")
            continue
        del seats[seatno]
        bookedseats -= 1
        print("Booking cancelled successfully.")
        print("Available Seats:",
              total_seats - bookedseats)
    elif choice == 3:

        if bookedseats == 0:
            print("No bookings available.")
            continue
        current_time = datetime.now()
        view_limit = current_time + timedelta(hours=view_n)
        if view_limit > travel_date:
            print("\nView booking is not allowed.")
            print("View Booking n:", view_n, "hours")
            print("Travel date/time:",
                  travel_date.strftime("%d-%m-%Y %I:%M %p"))
            continue
        try:
            seatno = int(input("Enter your seat number: "))
        except ValueError:
            print("Invalid seat number.")
            continue
        if seatno < 1 or seatno > total_seats:
            print("Invalid seat number.")
            continue
        if seatno not in seats:
            print("This seat is not booked.")
            continue
        phone = input(
            "Enter your registered phone number: "
        ).strip()
        if phone != seats[seatno]["phone"]:
            print("Invalid phone number.")
            continue
        print("\n--- Booking Details ---")
        print("Name:", seats[seatno]["name"])
        print("Age:", seats[seatno]["age"])
        print("Phone:", seats[seatno]["phone"])
        print("Seat Number:", seatno)
        print("Travel Date/Time:",
              travel_date.strftime("%d-%m-%Y %I:%M %p"))
        last_seat = min(
            seatno + view_n,
            total_seats
        )
        print("\nSeats from current seat + n:")

        for i in range(seatno, last_seat + 1):
            if i in seats:
                print("Seat", i, "→ Booked")
            else:
                print("Seat", i, "→ Available")
    elif choice == 0:

        print("\nThank you for using the Seat Booking System.")
        break

    else:
        print(
            "Invalid choice. Please select 0, 1, 2 or 3."
        )
```
