```python
bookedseats = 0
seat1 = False
seat2 = False
while True:
    print("\n1. New Booking")
    print("2. Cancel Booking")
    print("3. View Booking")
    print("0. Quit")
    try:
        choice = int(input("Enter your choice: "))
    except ValueError:
        print("Invalid choice. Please enter numbers only.")
        continue
    if choice == 1:
        if bookedseats == 2:
            print("All seats are already booked.")
            continue
        try:
            seatno = int(input("Enter seat number (1 or 2): "))
        except ValueError:
            print("Invalid seat number. Please enter numbers only.")
            continue
        if seatno != 1 and seatno != 2:
            print("Invalid seat number.")
            continue
        if seatno == 1 and seat1:
            print("Seat 1 is already booked.")
            continue
        if seatno == 2 and seat2:
            print("Seat 2 is already booked.")
            continue
        name = input("Enter your name: ")
        if not name.replace(" ", "").isalpha():
            print("Invalid name. Use alphabets and spaces only.")
            continue
        try:
            age = int(input("Enter your age: "))
        except ValueError:
            print("Invalid age. Please enter numbers only.")
            continue
        if age <= 0:
            print("Invalid age. Age must be greater than 0.")
            continue
        phone = input("Enter your phone number: ")
        if len(phone) != 10 or not phone.isdigit():
            print("Invalid phone number.")
            print("Phone number must contain exactly 10 digits.")
            continue
        if phone[0] not in "6789":
            print("Invalid phone number.")
            print("Phone number must start with 6, 7, 8 or 9.")
            continue
        if seatno == 1:
            seat1 = True
            name1 = name
            age1 = age
            phone1 = phone
        else:
            seat2 = True
            name2 = name
            age2 = age
            phone2 = phone
        bookedseats += 1
        print("Booking successful!")
        print("Seat Number:", seatno)
        print("Available Seats:", 2 - bookedseats)
    elif choice == 2:
        if bookedseats == 0:
            print("No booking available to cancel.")
            continue
        try:
            seatno = int(input("Enter your seat number: "))
        except ValueError:
            print("Invalid seat number. Please enter numbers only.")
            continue
        if seatno != 1 and seatno != 2:
            print("Invalid seat number.")
            continue
        if seatno == 1 and not seat1:
            print("Seat 1 is not booked.")
            continue

        if seatno == 2 and not seat2:
            print("Seat 2 is not booked.")
            continue
        phone = input("Enter your registered phone number: ")
        if seatno == 1:
            if phone != phone1:
                print("Invalid phone number.")
                continue
        else:
            if phone != phone2:
                print("Invalid phone number.")
                continue
        if seatno == 1:
            seat1 = False
        else:
            seat2 = False
        bookedseats -= 1
        print("Booking cancelled successfully.")
        print("Available Seats:", 2 - bookedseats)
    elif choice == 3:
        if bookedseats == 0:
            print("No bookings available.")
            continue
        try:
            seatno = int(input("Enter your seat number: "))
        except ValueError:
            print("Invalid seat number. Please enter numbers only.")
            continue
        if seatno != 1 and seatno != 2:
            print("Invalid seat number.")
            continue
        if seatno == 1 and not seat1:
            print("Seat 1 is not booked.")
            continue
        if seatno == 2 and not seat2:
            print("Seat 2 is not booked.")
            continue
        phone = input("Enter your registered phone number: ")
        if seatno == 1:
            if phone != phone1:
                print("Invalid phone number.")
                continue
            print("\n--- Booking Details ---")
            print("Name:", name1)
            print("Age:", age1)
            print("Phone:", phone1)
            print("Seat Number: 1")
        else:
            if phone != phone2:
                print("Invalid phone number.")
                continue
            print("\n--- Booking Details ---")
            print("Name:", name2)
            print("Age:", age2)
            print("Phone:", phone2)
            print("Seat Number: 2")
    elif choice == 0:
        print("Thank you for using the Seat Booking System.")
        break
    else:
        print("Invalid choice. Please select 0, 1, 2 or 3.")
```
